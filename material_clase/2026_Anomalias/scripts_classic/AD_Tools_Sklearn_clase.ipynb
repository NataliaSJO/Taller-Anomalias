```python
import matplotlib.pyplot as plt
import numpy as np
```

## Datos

```python
# Leemos IRIS
from sklearn.datasets import load_iris
X, y = load_iris(return_X_y=True)

print(X.shape)
print(y.shape)
```

## Anomaly detection

```python
# ALGORITMO EN SKLEARN

# Leemos un algoritmo: IsolationForest detector
from sklearn.ensemble import IsolationForest
```

```python
# Creamos el modelo
clf = IsolationForest(n_estimators=10, warm_start=True)
# Entrenamos el algoritmo
clf.fit(X)
```

```python
# Evaluamos el modelo
# "The lower, the more abnormal." (por eso el menos '-')
y_train_scores = -clf.score_samples(X)
```

```python
y_train_scores
```

```python
# Mostramos los datos y el dato mas anólao para las dos primeras dimensiones
plt.plot(X[:,0],X[:,1],'.')

#ii = np.argmax(y_train_scores)
ii = y_train_scores>0.55
plt.plot(X[ii,0],X[ii,1],'r.')
```

```python
# Mostramos las combinaciones de dimensiones 2 a 2
plt.figure(figsize=(14,6))
plt.subplot(3,3,1), plt.plot(X[:,0],X[:,1],'.'), plt.title('Dimensiones 1 y 2')
plt.subplot(3,3,1), plt.plot(X[ii,0],X[ii,1],'r.'), plt.title('Dimensiones 1 y 2')
plt.axis('equal')
plt.subplot(3,3,2), plt.plot(X[:,0],X[:,2],'.'), plt.title('Dimensiones 1 y 3')
plt.subplot(3,3,2), plt.plot(X[ii,0],X[ii,2],'r.'), plt.title('Dimensiones 1 y 3')
plt.axis('equal')
plt.subplot(3,3,3), plt.plot(X[:,0],X[:,3],'.'), plt.title('Dimensiones 1 y 4')
plt.subplot(3,3,3), plt.plot(X[ii,0],X[ii,3],'r.'), plt.title('Dimensiones 1 y 4')
plt.axis('equal')
plt.subplot(3,3,5), plt.plot(X[:,1],X[:,2],'.'), plt.title('Dimensiones 2 y 3')
plt.subplot(3,3,5), plt.plot(X[ii,1],X[ii,2],'r.'), plt.title('Dimensiones 2 y 3')
plt.axis('equal')
plt.subplot(3,3,6), plt.plot(X[:,1],X[:,3],'.'), plt.title('Dimensiones 2 y 4')
plt.subplot(3,3,6), plt.plot(X[ii,1],X[ii,3],'r.'), plt.title('Dimensiones 2 y 4')
plt.axis('equal')
plt.subplot(3,3,9), plt.plot(X[:,2],X[:,3],'.'), plt.title('Dimensiones 3 y 4')
plt.subplot(3,3,9), plt.plot(X[ii,2],X[ii,3],'r.'), plt.title('Dimensiones 3 y 4')
plt.axis('equal')
```

```python
# Leemos un algoritmo: DBSCAN
from sklearn.cluster import DBSCAN

# Creamos el modelo
clf = DBSCAN(eps=0.5, min_samples=5)

# Entrenamos el algoritmo
clf.fit(X)
```

```python
# Etiquetas de clustering DBSCAN
y = clf.labels_
y
```

```python
# Solo en 2 dimensiones
plt.scatter(X[:,0],X[:,1],c=y)
```

```python
# En varias dimensiones
A = y

ccc = 'viridis'
plt.subplot(3,3,1), plt.scatter(X[:,0],X[:,1], s=1, c=A, cmap=ccc), plt.title('Dimensiones 1 y 2')
plt.subplot(3,3,2), plt.scatter(X[:,0],X[:,2], s=1, c=A, cmap=ccc), plt.title('Dimensiones 1 y 3')
plt.subplot(3,3,3), plt.scatter(X[:,0],X[:,3], s=1, c=A, cmap=ccc), plt.title('Dimensiones 1 y 4')
plt.subplot(3,3,5), plt.scatter(X[:,1],X[:,2], s=1, c=A, cmap=ccc), plt.title('Dimensiones 2 y 3')
plt.subplot(3,3,6), plt.scatter(X[:,1],X[:,3], s=1, c=A, cmap=ccc), plt.title('Dimensiones 2 y 4')
plt.subplot(3,3,9), plt.scatter(X[:,2],X[:,3], s=1, c=A, cmap=ccc), plt.title('Dimensiones 3 y 4');
```

```python
#############################
# Leemos un algoritmo: GMM
#############################
from sklearn.mixture import GaussianMixture

# Creamos el modelo
# cov_type in ["spherical", "diag", "tied", "full"]

estimator = GaussianMixture(n_components=2, covariance_type="full", max_iter=20, random_state=0)

# Entrenamos el algoritmo
estimator.fit(X)
```

```python
y = estimator.predict(X)
y
```

```python
a = estimator.bic(X)
a
```

```python
# Solo en 2 dimensiones
plt.scatter(X[:,0],X[:,1],c=y)
```

```python
# En varias dimensiones
A = y

ccc = 'viridis'
plt.subplot(3,3,1), plt.scatter(X[:,0],X[:,1], s=1, c=A, cmap=ccc), plt.title('Dimensiones 1 y 2')
plt.subplot(3,3,2), plt.scatter(X[:,0],X[:,2], s=1, c=A, cmap=ccc), plt.title('Dimensiones 1 y 3')
plt.subplot(3,3,3), plt.scatter(X[:,0],X[:,3], s=1, c=A, cmap=ccc), plt.title('Dimensiones 1 y 4')
plt.subplot(3,3,5), plt.scatter(X[:,1],X[:,2], s=1, c=A, cmap=ccc), plt.title('Dimensiones 2 y 3')
plt.subplot(3,3,6), plt.scatter(X[:,1],X[:,3], s=1, c=A, cmap=ccc), plt.title('Dimensiones 2 y 4')
plt.subplot(3,3,9), plt.scatter(X[:,2],X[:,3], s=1, c=A, cmap=ccc), plt.title('Dimensiones 3 y 4');
```
