```python

import requests
from io import BytesIO

# Download the .npz file from the URL
url = "https://raw.githubusercontent.com/Minqi824/ADBench/main/adbench/datasets/Classical/6_cardio.npz"
response = requests.get(url)
response.raise_for_status()

# Keep the BytesIO object open to access data later
f = BytesIO(response.content)
data = np.load(f)

# Print the available keys within the .npz file
print(data.files)

X = data['X']  # Replace with the actual key for the data
y = data['y']  # Replace with the actual key for the labels


# Close the BytesIO object
f.close()
```
