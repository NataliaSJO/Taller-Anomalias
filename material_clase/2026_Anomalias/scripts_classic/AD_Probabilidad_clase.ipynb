## Modelo: Datos aleatorios 1D

```python
# Modelo Gaussiano

u = np.mean(x) # media real
s = np.std(x) # desviación estandar real

y=(1/(np.sqrt(2*np.pi*s*s)))*np.exp(-(((x-u)**2)/(2*s*s)))

# Probabilidad en cada dato asumiendo modelo Gausiano
plt.plot(x,y,'.')
plt.plot(R[0:-1],H,'ro-')
```

```python
# Anomalias con histograma
plt.plot(x,1/np.log(pp),'r.')
# Anomalias asumiendo Gaussiana
plt.plot(x,1/np.log(y),'.')
```

## Muchos datos 2D

Encuentra anomalias en los siguientes conjuntos de datos 2D.

```python
N = 1000

X = (np.random.randn(N,2))-0.5

plt.plot(X[:,0],X[:,1],'.')
plt.axis('equal')
```

```python
mu = np.mean(X,axis=0)
```

```python
S = np.cov(X.T)
```

```python
from scipy.stats import multivariate_normal
pX = multivariate_normal.pdf(X, mean=mu, cov=S)
```

```python

plt.scatter(X[:,0],X[:,1], s=1, c=1/np.log(pX), cmap='Reds')
plt.xlabel('X')
plt.ylabel('Y')
plt.axis('equal')
```

```python
N = 1000

X2 = np.sign(np.random.randn(1000,2))*np.abs(np.random.randn(1000,2))**0.5

plt.plot(X2[:,0],X2[:,1],'.')
plt.axis('equal')
```

```python

pX2 = multivariate_normal.pdf(X2, mean=mu, cov=S)
```

```python
plt.scatter(X2[:,0],X2[:,1], s=1, c=1/np.log(pX2), cmap='Reds')
plt.xlabel('X')
plt.ylabel('Y')
plt.axis('equal')
```
