## Muchos datos 1D

```python
N = 1000

# Valores ejemplo
y = np.random.rand(N,1)

# Valores test
y_hat = y + 0.1*np.random.randn(y.shape[0],y.shape[1])
```

```python
plt.plot(y,y_hat,'.')
plt.xlabel('valores ejemplo')
plt.ylabel('valores test')
```

## Muchos datos 2D

```python
N = 1000

# Valores ejemplo
x = 2*((np.random.rand(N,1))-0.5)
y = x**2

# Valores test
sigma = 0.05
y_hat = y + sigma*np.random.randn(y.shape[0],y.shape[1])

plt.plot(x,y,'.')
plt.plot(x,y_hat,'r.')
plt.xlabel('X')
plt.ylabel('Y')
```
